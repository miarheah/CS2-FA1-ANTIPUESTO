```python
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print(f"Distance: {distance}")
```

```python
import math

r = float(input("Enter the radius: "))

area = math.pi * (r ** 2)
circumference = 2 * math.pi * r

print(f"Area: {area}")
print(f"Circumference: {circumference}")
```

```python
import math

a = float(input("Enter leg a: "))
b = float(input("Enter leg b: "))

# Method 1: Using math.sqrt() and exponentiation
hypot1 = math.sqrt(a ** 2 + b ** 2)

# Method 2: Using math.hypot() shortcut
hypot2 = math.hypot(a, b)

print(f"Hypotenuse (using math.sqrt): {hypot1}")
print(f"Hypotenuse (using math.hypot): {hypot2}")
```
